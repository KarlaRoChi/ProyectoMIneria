Descarga el **Stroke Prediction Dataset** desde Kaggle usando `kagglehub` y realiza una revisión inicial de atributos, tipos de datos y calidad.

> **Requisito:** tener configurado `~/.kaggle/kaggle.json`  
> O colocar el CSV manualmente en `data/healthcare-dataset-stroke-data.csv`

```python
import os
import shutil
import pandas as pd

DATA_DIR = os.path.join(os.getcwd(), "data")
CSV_NAME = "healthcare-dataset-stroke-data.csv"
CSV_PATH = os.path.join(DATA_DIR, CSV_NAME)

os.makedirs(DATA_DIR, exist_ok=True)
print("DATA_DIR:", DATA_DIR)
```

## 1.1 Descarga del dataset

```python
if os.path.exists(CSV_PATH):
    print(f"[INFO] Dataset ya presente en: {CSV_PATH}")
else:
    import kagglehub
    path = kagglehub.dataset_download("fedesoriano/stroke-prediction-dataset")
    for root, _, files in os.walk(path):
        for fname in files:
            if fname.endswith(".csv"):
                shutil.copy(os.path.join(root, fname), CSV_PATH)
                print(f"[INFO] Dataset copiado a: {CSV_PATH}")
                break
```

## 1.2 Carga del CSV

```python
df = pd.read_csv(CSV_PATH)
print(f"Filas: {df.shape[0]}  |  Columnas: {df.shape[1]}")
df.head()
```

## 1.3 Tipos de datos

```python
df.dtypes
```

## 1.4 Valores faltantes

```python
missing = df.isnull().sum()
missing[missing > 0].rename("valores faltantes").to_frame()
```

## 1.5 Distribución de la variable objetivo (`stroke`)

```python
counts = df["stroke"].value_counts()
pct    = df["stroke"].value_counts(normalize=True) * 100
pd.DataFrame({"count": counts, "pct (%)": pct.round(2)})
```
